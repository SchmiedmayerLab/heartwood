Terra supports the Heartwood terminal and notebook interfaces. Keep the terminal process that supervises a Heartwood-managed model running, and use a different session identifier in this notebook.

Submit a bounded, read-only first task. The call returns after the background OpenHands run starts. Any proposed tools appear in `pending_approval` and are not executed until the complete action set is allowed.

```python
view = session.chat(
    "Inspect the files in this project and summarize their purpose. "
    "Do not modify files or access parent directories."
)
view.lifecycle.status
```

```python
import time

while view.lifecycle.status == "running":
    time.sleep(0.5)
    view = session.replay()

[(message.label, message.content) for message in view.conversation]
```

If `view.pending_approval` is present, inspect every `actions` member. Resolve the complete set with `session.approve(group_id=view.pending_approval.group_id)` or reject it with `session.deny(group_id=...)`.
